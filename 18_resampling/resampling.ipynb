```text
================================================================================
Experiment 18: Bootstrapping and Cross-Validation on Iris Dataset
================================================================================

AIM:
Implement bootstrapping and cross-validation on the Iris dataset. Compare the model performance metrics (accuracy, F1-score) obtained using these resampling methods. Discuss the advantages and disadvantages of each method.

ALGORITHM / PROCEDURE:
Part A: Bootstrapping with Out-Of-Bag (OOB) Testing
Step 1: For each of 10 iterations:
   - Generate bootstrap sample with replacement: (X_boot, y_boot).
   - Identify unpicked out-of-bag (OOB) samples as test set: (X_test, y_test).
   - Fit DecisionTreeClassifier(random_state=42) on bootstrap training set.
   - Predict OOB test set, compute Accuracy and weighted F1-score.
Step 2: Compute mean Bootstrapping Accuracy and F1-score.

Part B: 5-Fold Cross-Validation
Step 1: Partition dataset into 5 equal folds using KFold(n_splits=5, shuffle=True, random_state=42).
Step 2: For each fold, hold out fold as test set and train on remaining 4 folds.
Step 3: Predict test set, record Accuracy and weighted F1-score.
Step 4: Compute mean 5-Fold Cross-Validation Accuracy and F1-score.
Step 5: Compare resampling methods regarding bias, variance, and sample reuse.

SAMPLE OUTPUT / METRICS:
--- Bootstrapping Results (10 Iterations) ---
Average Accuracy : 0.9452
Average F1 Score : 0.9450

--- 5-Fold Cross-Validation Results ---
Average Accuracy : 0.9533
Average F1 Score : 0.9533

--- Summary Comparison ---
                 Method  Average Accuracy  Average F1 Score
    Bootstrapping (OOB)            0.9452            0.9450
5-Fold Cross-Validation            0.9533            0.9533
```

````
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import KFold
from sklearn.utils import resample

# 1. Load Dataset
iris = load_iris()
X = iris.data
y = iris.target
model = DecisionTreeClassifier(random_state=42)

# 2. Bootstrapping with OOB Evaluation
boot_accuracy = []
boot_f1 = []
n_iterations = 10

for i in range(n_iterations):
    indices = resample(range(len(X)), replace=True, random_state=i)
    X_boot, y_boot = X[indices], y[indices]
    
    oob = list(set(range(len(X))) - set(indices))
    if len(oob) == 0:
        continue
    X_test, y_test = X[oob], y[oob]
    
    model.fit(X_boot, y_boot)
    pred = model.predict(X_test)
    boot_accuracy.append(accuracy_score(y_test, pred))
    boot_f1.append(f1_score(y_test, pred, average='weighted'))

print("--- Bootstrapping Results (10 Iterations) ---")
print(f"Average Accuracy : {np.mean(boot_accuracy):.4f}")
print(f"Average F1 Score : {np.mean(boot_f1):.4f}")

# 3. 5-Fold Cross Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)
cv_accuracy = []
cv_f1 = []

for train_index, test_index in kf.split(X):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]
    
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    cv_accuracy.append(accuracy_score(y_test, pred))
    cv_f1.append(f1_score(y_test, pred, average='weighted'))

print("\n--- 5-Fold Cross-Validation Results ---")
print(f"Average Accuracy : {np.mean(cv_accuracy):.4f}")
print(f"Average F1 Score : {np.mean(cv_f1):.4f}")

# 4. Resampling Comparison
res_df = pd.DataFrame({
    'Method': ['Bootstrapping (OOB)', '5-Fold Cross-Validation'],
    'Average Accuracy': [round(np.mean(boot_accuracy), 4), round(np.mean(cv_accuracy), 4)],
    'Average F1 Score': [round(np.mean(boot_f1), 4), round(np.mean(cv_f1), 4)]
})
print("\n--- Summary Comparison ---")
print(res_df.to_string(index=False))
````
