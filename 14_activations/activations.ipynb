```text
================================================================================
Experiment 14: Neural Network using Different Activation Functions on MNIST Dataset
================================================================================

AIM:
Implement and compare the performance of a neural network using different activation functions (Sigmoid, ReLU, Tanh) on the MNIST dataset. Analyze how each activation function affects the training process and classification accuracy.

ALGORITHM / PROCEDURE:
Step 1: Load MNIST digits dataset, normalize pixels to [0, 1], and flatten images into 784-dimensional vectors.
Step 2: One-hot encode digit labels (0-9).
Step 3: Build feed-forward neural network architecture: Input(784) -> Hidden1(128) -> Hidden2(64) -> Output(10).
Step 4: For each activation function in ['sigmoid', 'relu', 'tanh']:
   - Set hidden layer activations.
   - Train the network and record total training time.
   - Evaluate test accuracy and loss.
Step 5: Compare convergence speed and accuracy across activation functions.

SAMPLE OUTPUT / METRICS:
Training Neural Networks with Different Activations:
Activation: Sigmoid (logistic)   | Accuracy: 96.67% | Training Time: 0.85s
Activation: ReLU                 | Accuracy: 97.78% | Training Time: 0.65s
Activation: Tanh                 | Accuracy: 97.50% | Training Time: 0.70s

--- Activation Function Comparison Summary ---
Activation Function  Test Accuracy  Training Time (s)
 Sigmoid (logistic)         0.9667               0.85
               ReLU         0.9778               0.65
               Tanh         0.9750               0.70
```

````
import time
import numpy as np
import pandas as pd
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

# 1. Load Dataset (MNIST digits)
print("Loading Digits dataset...")
digits = load_digits()
X = digits.data
y = digits.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# 2. Compare Activation Functions
activation_functions = [
    ('Sigmoid (logistic)', 'logistic'),
    ('ReLU', 'relu'),
    ('Tanh', 'tanh')
]

results = []
print("\nTraining Neural Networks with Different Activations:")
for display_name, act in activation_functions:
    model = MLPClassifier(
        hidden_layer_sizes=(128, 64),
        activation=act,
        solver='adam',
        max_iter=300,
        random_state=42
    )
    
    start_time = time.time()
    model.fit(X_train, y_train)
    training_time = time.time() - start_time
    
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    
    results.append([display_name, round(acc, 4), round(training_time, 2)])
    print(f"Activation: {display_name:20s} | Accuracy: {acc*100:.2f}% | Training Time: {training_time:.2f}s")

# 3. Summary
res_df = pd.DataFrame(results, columns=['Activation Function', 'Test Accuracy', 'Training Time (s)'])
print("\n--- Activation Function Comparison Summary ---")
print(res_df.to_string(index=False))
````
