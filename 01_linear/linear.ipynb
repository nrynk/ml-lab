```text
================================================================================
Experiment 1: Linear Regression with One Variable on California Housing Dataset
================================================================================

AIM:
To implement Linear Regression with one independent variable using the California Housing dataset and predict housing prices based on the average number of rooms per dwelling (AveRooms) using:
1. Gradient Descent
2. Normal Equation
and evaluate the model using Mean Squared Error (MSE) and R² Score.

ALGORITHM / PROCEDURE:
Method 1: Linear Regression using Gradient Descent
Step 1: Import required libraries (numpy, pandas, matplotlib, sklearn).
Step 2: Load the California Housing dataset using fetch_california_housing.
Step 3: Select Feature (X): Average number of rooms (AveRooms), Target (y): Median house value.
Step 4: Normalize the feature values using StandardScaler for faster convergence.
Step 5: Initialize parameters: theta0 (bias), theta1 (slope), learning_rate (0.01), iterations (1000).
Step 6: Repeat until convergence:
   - Compute predictions: y_pred = theta0 + theta1 * X
   - Compute gradients: d_theta0 = (1/m)*sum(y_pred - y), d_theta1 = (1/m)*sum((y_pred - y)*X)
   - Update: theta0 = theta0 - alpha * d_theta0, theta1 = theta1 - alpha * d_theta1
Step 7: Compute Mean Squared Error (MSE) and R² score for Gradient Descent.

Method 2: Linear Regression using Normal Equation
Step 1: Add a bias column of ones to feature matrix X: X_normal = [1, X].
Step 2: Compute parameters analytically: theta = inv(X_normal.T @ X_normal) @ X_normal.T @ y.
Step 3: Predict housing prices: y_pred_ne = X_normal @ theta.
Step 4: Compute MSE and R² score for Normal Equation.
Step 5: Plot actual data scatter and fitted regression line. Save plot to linear_plot.png.

SAMPLE OUTPUT / METRICS:
--- Gradient Descent Results ---
Intercept (theta0): 2.0686
Slope (theta1):     0.3182
MSE:                1.2789
R2 Score:           0.0238

--- Normal Equation Results ---
Theta Values (Intercept, Slope): [1.7495, 0.0676]
MSE:                             1.2783
R2 Score:                        0.0243

Plot saved successfully to 'linear_plot.png'.
```

````
import os
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

# 1. Load Dataset
print("Loading California Housing dataset...")
housing = fetch_california_housing(as_frame=True)
X = housing.data[['AveRooms']].values
y = housing.target.values

# 2. Feature Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Linear Regression using Gradient Descent
m = len(y)
theta0 = 0.0
theta1 = 0.0
learning_rate = 0.01
iterations = 1000

for i in range(iterations):
    y_pred = theta0 + theta1 * X_scaled.flatten()
    d_theta0 = (1/m) * np.sum(y_pred - y)
    d_theta1 = (1/m) * np.sum((y_pred - y) * X_scaled.flatten())
    theta0 -= learning_rate * d_theta0
    theta1 -= learning_rate * d_theta1

y_pred_gd = theta0 + theta1 * X_scaled.flatten()
mse_gd = mean_squared_error(y, y_pred_gd)
r2_gd = r2_score(y, y_pred_gd)

print("\n--- Gradient Descent Results ---")
print(f"Intercept (theta0): {theta0:.4f}")
print(f"Slope (theta1):     {theta1:.4f}")
print(f"MSE:                {mse_gd:.4f}")
print(f"R2 Score:           {r2_gd:.4f}")

# 4. Linear Regression using Normal Equation
X_normal = np.c_[np.ones((m, 1)), X]
theta = np.linalg.inv(X_normal.T.dot(X_normal)).dot(X_normal.T).dot(y)
y_pred_ne = X_normal.dot(theta)
mse_ne = mean_squared_error(y, y_pred_ne)
r2_ne = r2_score(y, y_pred_ne)

print("\n--- Normal Equation Results ---")
print(f"Theta Values (Intercept, Slope): [{theta[0]:.4f}, {theta[1]:.4f}]")
print(f"MSE:                             {mse_ne:.4f}")
print(f"R2 Score:                        {r2_ne:.4f}")

# 5. Visualization
plt.figure(figsize=(8, 6))
plt.scatter(X, y, alpha=0.3, label='Actual Data', color='royalblue')
sorted_index = np.argsort(X.flatten())
plt.plot(
    X[sorted_index],
    y_pred_ne[sorted_index],
    color='red',
    linewidth=2,
    label='Regression Line (Normal Eq)'
)
plt.xlabel("Average Rooms (AveRooms)")
plt.ylabel("Median House Value ($100,000s)")
plt.title("Linear Regression on California Housing Dataset")
plt.legend()
plt.tight_layout()
plt.savefig("linear_plot.png", dpi=150)
print("\nPlot saved successfully to 'linear_plot.png'.")
````
